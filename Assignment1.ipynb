Calculates the difference of the records with the previous records using [diff function](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.diff.html#pandas-dataframe-diff) for `thal` column. Then, plot column `thal` with a histogram.

```python
heart_disease['thal'] = heart_disease['thal'].astype(float)
```

```python
heart_disease['thal_diff'] = heart_disease['thal'].diff()
```

```python
ax = heart_disease['thal_diff'].plot(
    kind='hist',
    bins=20,
    figsize=(10, 6),
    color='purple',
    alpha=0.75,
    title='Difference in Thal Values (Using diff())'
)

ax.set_xlabel('Thal Difference')
ax.set_ylabel('Frequency')
ax.grid(True, linestyle='--', alpha=0.4)
```

```python
import os

def save_figures(fig_list, folder="data"):
    os.makedirs(folder, exist_ok=True)
    for i, fig in enumerate(fig_list, start=1):
        fig.savefig(f"{folder}/figure_{i}.png")
        print(f"Saved figure_{i}.png")
```

```python
save_figures(plots)
```
